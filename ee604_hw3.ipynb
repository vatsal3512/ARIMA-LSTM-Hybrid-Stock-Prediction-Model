```python
import cv2
import numpy as np
import os
from google.colab import files


IMAGE_PATH = 'grayscaleimg.jpg'
OUTPUT_DIR = 'bit_plane_output'
VIDEO_NAME = 'bit_plane_reconstruction.mp4'


if not os.path.exists(OUTPUT_DIR):
    os.makedirs(OUTPUT_DIR)

print(f"Attempting to load image from: {IMAGE_PATH}")
original_image = cv2.imread(IMAGE_PATH, cv2.IMREAD_GRAYSCALE)

if original_image is None:
    print(f"--- ERROR ---")
    print(f"Could not open or find the image at '{IMAGE_PATH}'.")
    print("Please make sure the image was uploaded correctly to the Colab session.")
    exit()

print("Image loaded successfully.")
image_files = []
reconstructed_image = np.zeros(original_image.shape, dtype=np.uint8)
print("Starting bit-plane slicing and reconstruction...")
for bit in range(7, -1, -1):
    plane = np.bitwise_and(original_image, 2**bit)
    reconstructed_image = np.add(reconstructed_image, plane)
    filename = os.path.join(OUTPUT_DIR, f'reconstructed_with_planes_8_to_{bit+1}.png')
    cv2.imwrite(filename, reconstructed_image)
    image_files.append(filename)
    print(f"Generated image with bit planes 8 down to {bit+1}.")

print(f"Creating video: {VIDEO_NAME}...")
frame = cv2.imread(image_files[0])
height, width, layers = frame.shape
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
video = cv2.VideoWriter(os.path.join(OUTPUT_DIR, VIDEO_NAME), fourcc, 1/1.5, (width, height))
for image_file in image_files:
    video.write(cv2.imread(image_file))
video.release()
print(f"\nVideo successfully created at '{os.path.join(OUTPUT_DIR, VIDEO_NAME)}'")

video_path_to_download = os.path.join(OUTPUT_DIR, VIDEO_NAME)
files.download(video_path_to_download)
```

```python
import os
from PIL import Image

# pdf generation
image_folder = os.path.join(OUTPUT_DIR)
image_files = sorted([f for f in os.listdir(image_folder) if f.endswith('.png')])

first_image_path = os.path.join(image_folder, image_files[0])
img_list = [Image.open(first_image_path).convert('RGB')]

for filename in image_files[1:]:
    img_path = os.path.join(image_folder, filename)
    img = Image.open(img_path).convert('RGB')
    img_list.append(img)


pdf_path = os.path.join(image_folder, 'bit_plane_reconstruction.pdf')

img_list[0].save(pdf_path, save_all=True, append_images=img_list[1:])

print(f"\nPDF successfully created at '{pdf_path}'")

files.download(pdf_path)
```
